```python
from pyspark import pipelines as dp

@dp.table(name="inzentio.bronze.product")
def bronze_product():
    return spark.read.table("inzentio.event_log.product")
```

```python
from pyspark import pipelines as dp

@dp.table(name="inzentio.silver.product_curated")

def product_curated():
    return (
        spark.read.table("inzentio.bronze.product")
        .select(
            "ProductID",
            "Name",
            "ProductNumber",
            "Color",
            "ListPrice",
            "ModifiedDate"
        )
    )
```

```python
from pyspark import pipelines as dp

from pyspark.sql.functions import avg, count, round

@dp.table(name="inzentio.gold.product_summary")
def product_summary():
    return (
        spark.read.table("inzentio.silver.product_curated")
        .groupBy("Color")
        .agg(
            count("*").alias("TotalProducts"),
            round(avg("ListPrice"), 3).alias("AveragePrice")
        )
    )
```
